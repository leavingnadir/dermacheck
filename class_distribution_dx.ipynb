```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/raw/HAM10000_metadata.csv')  # adjust path if needed


counts = df['dx'].value_counts()
plt.figure(figsize=(8,5))
counts.plot(kind='bar', color='steelblue')
plt.title('Class Distribution of Diagnosis (dx)')
plt.xlabel('Diagnosis Class')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('dx_class_distribution.png')
plt.show()
```
